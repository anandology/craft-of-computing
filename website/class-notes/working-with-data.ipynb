````python
[n*n for n in numbers]
````

List comprehensions are written like this:

```python
[expr for var in some_list]

[expr for var in some_list if some_condition]
```

They are equivalant to:

```python
result = []
for var in some_list:
    result.append(expr)

result = []
for var in some_list:
    if some_condition:
        result.append(expr)

```

### Iterating over two lists together

````python
names = ["Alice", "Bob", "Charlie"]
marks = [10, 20, 30]
````

A naive way to do this would be:

````python
print(names[0], marks[0])
print(names[1], marks[1])
print(names[2], marks[2])
````

````python
n = len(names)
````

````python
for i in range(n):
    print(names[i], marks[i])
````

````python
for name, score in zip(names, marks):
    print(name, score)
````

````python
list(zip(names, marks))
````

Find all students who scored more than 15 marks.

````python
for name, score in zip(names, marks):
    if score > 15:
        print(name)
````

How to get the names of students as a list?

````python
[name for name, score in zip(names, marks) if score > 15]
````

````python
names_15 = [name for name, score in zip(names, marks) if score > 15]
````

````python
len(names_15)
````

````python
# percentage
len(names_15) / len(names) * 100
````

### Greeting in many languages

````python
def hello(name):
    print("Hello", name)
````

````python
hello("Python")
````

How to extend this program to greet in multiple languages.

A naive way would be to use an if condition.

````python
def hello(name, lang):
    if lang == "hi":
        print("Namaste", name)
    elif lang == "kn":
        print("Namaskara", name)
    else:
        print("Hello", name)
````

````python
hello("Python", "kn")
````

A better way to do this using a dictionary. 

````python
greetings = {
    "hi": "Namaste",
    "kn": "Namakara",
    "en": "Hello"
}

def hello(name, lang):
    prefix = greetings[lang]
    print(prefix, name)
````

````python
hello("Python", "kn")
````
