Let's see how to add options to our program.

````python
%%file hello.py
import argparse

p = argparse.ArgumentParser()
p.add_argument("name")
p.add_argument("-r", "--repeats", type=int, default=1, 
               help="number of times to repeat the message")

args = p.parse_args()

print(args)

for i in range(args.repeats):
    print("Hello", args.name)
````

Let's add `-u` or `--upper-case` flag to our hello.py.

````python
%%file hello.py
import argparse

p = argparse.ArgumentParser()
p.add_argument("name")
p.add_argument("-r", "--repeats", type=int, default=1, 
               help="number of times to repeat the message")
p.add_argument("-u", "--upper-case", action="store_true",
               help="convert the message to uppercase")

args = p.parse_args()

import sys
print(sys.argv)
print(args)

message = "hello " + args.name
if args.upper_case:
    message = message.upper()

for i in range(args.repeats):
    print(message)
````

````python
!python hello.py craft
````

````python
!python hello.py -u craft
````

````python
!python hello.py -u craft -r 2
````

````python
!python hello.py -r2  -u craft
````
