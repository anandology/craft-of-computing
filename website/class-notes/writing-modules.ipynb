### The `__name__` magic variable

That is `name` with two underscores on both sides.

````python
%%file mymodule2.py

x = 2

def add(a, b):
    return a+b

print(__name__)
print(add(3, 4))
````

````python
!python mymodule2.py
````

````python
import mymodule2
````

Now we can use `__name__` to decide whether a file is run as a script or imported as a module.

````python
%%file mymodule3.py

x = 2

def add(a, b):
    return a+b

# Run the following code only when this file is run as as script
# Ignore it when this is imported as a module
if __name__ == "__main__":
    print(add(3, 4))
````

````python
!python mymodule3.py
````

````python
import mymodule3
````

````python
mymodule3.add(3, 4)
````

### Adding docstrings to a module

````python
%%file sq.py
"""
The square module.

The square module provides a function to compute square of a number.

This can also be used as a script.

USAGE:

    $ python sq.py 5
    25
"""
import sys

def square(x):
    """Computes square of a number.

        >>> square(3)
        9
    """
    return x*x

def main():
    n = int(sys.argv[1])
    print(square(n))

if __name__ == "__main__":
    main()
````

````python
# reload sq module
import importlib
importlib.reload(sq)
````

````python
help(sq)
````

````python
%%file square.py

def square(x):
    return x*x

def sum_of_squares(x, y):
    return square(x) + square(y)
````
