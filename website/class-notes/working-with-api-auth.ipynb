### Query Parameters

```python
!curl "https://figlet.apps.pipal.in/api/figlet?text=Python"
```

```python
url = "https://figlet.apps.pipal.in/api/figlet?text=Python"
response = requests.get(url)
print(response.text)
```

If the text has space or special characters, that need to be encoded properly. The request module can take care of that. 

```python
url = "https://figlet.apps.pipal.in/api/figlet"
params = {"text": "Python & Web"}
response = requests.get(url, params=params)
print(response.text)
```

```python
def figlet(text):
    url = "https://figlet.apps.pipal.in/api/figlet"
    params = {"text": text}
    response = requests.get(url, params=params)
    print(response.text)
```

```python
figlet("one & two")
```

```python
for i in range(5):
    figlet(str(i))
```

## Working with HTTP POST method

```python
url = "https://numbers.apps.pipal.in/product"
data = {"numbers": [1, 2, 3, 4, 5]}
response = requests.post(url, json=data)
d = response.json()
d
```

```python
d['result']
```

```python
def product(numbers):
    url = "https://numbers.apps.pipal.in/product"
    data = {"numbers": numbers}
    response = requests.post(url, json=data)
    d = response.json()
    return d['result']
```

```python
product([1, 2, 3])
```

## Authentication

```python
url = "https://numbers.apps.pipal.in/store"
```

```python
url = "https://numbers.apps.pipal.in/store"
token = "anand-token-123"

headers = {
    "Authorization": f"Bearer {token}"
}

response = requests.get(url, headers=headers)
print(response.status_code)
response.json()
```

There are no entries. Let's add a couple of them.

```python
url = "https://numbers.apps.pipal.in/store/x"
token = "anand-token-123"

headers = {
    "Authorization": f"Bearer {token}"
}
data = {"value": 42}
response = requests.post(url, json=data, headers=headers)
print(response.status_code)
response.json()
```
