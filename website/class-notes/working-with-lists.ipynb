Let's write a program `greet-all.py` to greet all people whos names are specified as command-line arguments.

```
! python greet-all.py Alice Bob Charlie

Hello Alice
Hello Bob
Hello Charlie
```

````python
%%file greet-all.py
import sys

args = sys.argv[1:]
print(args)

for name in args:
    print("Hello", name)
````

How to convert all the cities into uppercase?

````python
cities.upper() # this doesn't work. upper is a method on strings, not lists
````

````python
# this is just printing
for city in cities:
    print(city.upper())
````

But, I want the result as a list.

````python
upper_cities = []
for city in cities:
    upper_cities.append(city.upper())
    print(upper_cities)
````
