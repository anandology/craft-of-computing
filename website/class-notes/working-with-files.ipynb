Assume the list has 5 elements, this is how you can index them.


```
+----+----+----+----+----+
|  a |  b |  c |  d |  e |
+----+----+----+----+----+
|  0 |  1 |  2 |  3 |  4 | --> regular index
+----+----+----+----+----+
| -5 | -4 | -3 | -2 | -1 | <-- negative index
+----+----+----+----+----+
```

### List Slicing

````python
names = ["a", "b", "c", "d", "e", "f", "g", "h"]
````

````python
len(names)
````

````python
names[0:2] # from index 0 to index 2 (end not included)
````

````python
names[:2] # upto index 2
````

````python
names[2:]
````

````python
names[1:6]
````

````python
names[-2:] # last 2 elements
````

````python
names[1:7:2]
````

````python
names[:] # copy of the list
````

````python
names[::2] # every second element from the beginning to end
````

````python
names[::-1] 
````

### Example: Computing sum

````python
numbers = [1, 2, 3, 4, 5]
````

````python
sum(numbers)
````

````python
result = 0
for n in numbers:
    result = result + n
    print(n, result)

print(result)
````

````python
def my_sum(numbers):
    result = 0
    for n in numbers:
        result = result + n
    return result
    
````

````python
my_sum(numbers)
````
