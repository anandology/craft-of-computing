```python
url = "https://coc.apucomputing.in/rajdhani/ocean-tides/api/progress?status=open"

progress = requests.get(url).json()

for task in progress['tasks']:
    print(task['status'], task['title'])
```

```python
url = "https://coc.apucomputing.in/rajdhani/ocean-tides/api/progress"

# params are passed as a dictionary and requests takes care of encoding it
params = {"status": "open"}
progress = requests.get(url, params=params).json()
    
for task in progress['tasks']:
    print(task['status'], task['title'])
```

### Login API

```python
url = "https://coc.apucomputing.in/rajdhani-bookings/accounts/login"

site = "group-00"
# email = "demo@example.com"
# password = "rajdhani123"
email = "anand@example.com"
password = "anand123"
```

```python
data = {
    "site": site,
    "email": email,
    "password": password
}
```

```python
response = requests.post(url, json=data)
```

```python
response.status_code
```

```python
response.json()
```
